```
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
import os

# ── CONFIG ────────────────────────────────────────────────────────────────────
BASE_PATH   = '/content/drive/MyDrive/Fockup Dataset MIMIC IV/MIMIC IV 3.1'
INPUT_DIR   = f'{BASE_PATH}/fl_clients'
OUTPUT_DIR  = f'{BASE_PATH}/fl_clients/preprocessed'
os.makedirs(OUTPUT_DIR, exist_ok=True)

CLIENTS = ['client0_medical', 'client1_surgical', 'client2_cardiac']

# Vasopressor columns — binary, do NOT scale
VASO_COLS = ['vaso_dopamine','vaso_epinephrine','vaso_norepinephrine',
             'vaso_phenylephrine','vaso_vasopressin']

for client_name in CLIENTS:
    print(f"\nProcessing {client_name}...")

    # ── Load ──────────────────────────────────────────────────────────────────
    df = pd.read_csv(f'{INPUT_DIR}/{client_name}.csv')
    print(f"  Raw shape: {df.shape}")

    # ── Fill missing values ───────────────────────────────────────────────────
    df = df.fillna(0)
    print(f"  NaN after fill: {df.isna().sum().sum()}")

    # ── Scale all columns except label and vasopressors ───────────────────────
    target_col  = 'mortality'
    scale_cols  = [c for c in df.columns
                   if c != target_col
                   and c not in VASO_COLS]

    scaler      = StandardScaler()
    df[scale_cols] = scaler.fit_transform(df[scale_cols])

    # ── Save ──────────────────────────────────────────────────────────────────
    out_path = f'{OUTPUT_DIR}/{client_name}_preprocessed.csv'
    df.to_csv(out_path, index=False)

    print(f"  Scaled features : {len(scale_cols)}")
    print(f"  Binary features : {len(VASO_COLS)} (not scaled)")
    print(f"  Saved → {out_path}")

print("\nAll 3 clients preprocessed ✓")
```
